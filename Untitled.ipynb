```python
class Dataset_handler():

    def __init__(self, sequence, lidar=True, progress_bar=True, low_memory=True):

        self.lidar = lidar
        self.low_memory = low_memory

        self.seq_dir = '../dataset/sequences/{}/'.format(sequence)
        self.poses_dir = '../dataset/poses/{}.txt'.format(sequence)
        poses = pd.read_csv(self.poses_dir, delimiter = ' ', header=None)

        self.left_image_files = os.listdir(self.seq_dir + 'image_0')
        self.right_image_files = os.listdir(self.seq_dir + 'image_1')
        self.velodyne_files = os.listdir(self.seq_dir + 'velodyne')
        self.num_frames = len(self.left_image_files)
        self.lidar_path = self.seq_dir + 'velodyne/'

        poses = pd.read_csv(self.poses_dir, delimiter=" ",header=None)
        self.gt = np.zeros((self.num_frames, 3, 4))

        for i in range(len(poses)):
            self.gt[i] = np.array(poses.iloc[i]).reshape(3, 4)
        
        calib = pd.read_csv(self.seq_dir + 'calib.txt', delimiter=' ', header=None, index_col=0)
        self.P0 = np.array(calib.loc['P0:']).reshape((3, 4))
        self.P1 = np.array(calib.loc['P1:']).reshape((3, 4))
        self.P2 = np.array(calib.loc['P2:']).reshape((3, 4))
        self.P3 = np.array(calib.loc['P3:']).reshape((3, 4))
        self.Tr = np.array(calib.loc['Tr:']).reshape((3, 4))

        if low_memory:

            self.reset_frame()
            self.first_image_left = cv2.imread(self.seq_dir + 'image_0/'
                                              + self.left_image_files[0], 0)
            self.first_image_right = cv2.imread(self.seq_dir + 'image_1/'
                                              + self.left_image_files[0], 0)
            self.second_image_left = cv2.imread(self.seq_dir + 'image_0/' 
                                   + self.left_image_files[1], 0)
            if self.lidar:
                self.first_pointcloud = np.fromfile(self.lidar_path + self.velodyne_files[0],
                                                    dtype=np.float32, 
                                                    count=-1).reshape((-1, 4))
            self.imheight = self.first_image_left.shape[0]
            self.imwidth = self.first_image_left.shape[1]
            
        else:
            self.images_left  = []
            self.images_right = []
            self.pointclouds = []
            if progress_bar:
                import progressbar
                bar = progressbar.ProgressBar(max_value=self.num_frames)

            for i, image_left in enumerate(self.left_image_files):
                name_right = self.right_image_files[i]
                name_left = self.right_image_files[i]
                self.images_left.append(cv2.imread(self.seq_dir + 'image_0/' + name_left, 0))
                self.images_right.append(cv2.imread(self.seq_dir + 'image_1/' + name_right, 0))
                if self.lidar:
                    pointcloud = np.fromfile(self.lidar_path + self.velodyne_files[i], 
                                             dtype=np.float32, 
                                             count=-1).reshape([-1,4])
                    self.pointclouds.append(pointcloud)
                if progress_bar:
                    bar.update(i+1)
            
            self.imheight = self.images_left[0].shape[0]
            self.imwidth = self.images_left[0].shape[1]
            self.first_image_left = self.images_left[0]
            self.first_image_right = self.images_right[0]
            self.second_image_left = self.images_left[1]
            if self.lidar:
                self.first_pointcloud = self.pointclouds[0]

    def reset_frame(self):

        self.images_left = (cv2.imread(self.seq_dir + 'image_0/' + name_left, 0)
                           for name_left in self.left_image_files)

        self.images_right = (cv2.imread(self.seq_dir + 'image_1/' + name_left, 0)
                           for name_left in self.left_image_files)


        if self.lidar:
            self.pointclouds = (np.fromfile(self.lidar_path + velodyne_file, 
                                            dtype=np.float32, 
                                            count=-1).reshape((-1, 4))
                                for velodyne_file in self.velodyne_files)

    pass
```

```python
def compute_left_disparity_map(img_left, img_right, matcher='bm', rgb=False, verbose=False):

    sad_window = 6
    num_disparities = sad_window * 16
    block_size = 11
    matcher_name = matcher

    if matcher_name == 'bm':
        matcher = cv2.StereoBM_create(numDisparities=num_disparities,
                                     blockSize=block_size)
    elif matcher == 'sgbm':
        matcher =cv2.StereoSGBM_create(numDisparities=num_disparities,
                                      minDisparity=0,
                                      blockSize=block_size,
                                      P1 = 8*1 * sad_window ** 2,
                                      P2 = 32 * 1 * block_size ** 2,
                                      mode=cv2.STEREO_SGBM_MODE_SGBM_3WAY)

    if rgb:
        img_left = cv2.cvtColor(img_left, cv2.COLOR_BGR2GRAY)
        img_right = cv2.cvtColor(img_right, cv2.COLOR_BGR2GRAY)

    start = datetime.datetime.now()
    disp_left = matcher.compute(img_left, img_right).astype(np.float32)/16
    end = datetime.datetime.now()

    if verbose:
        print(f'Time to compute disparity map using Stereo{matcher_name.upper()}', end - start)

    return disp_left
        
    
```

```python
def decompose_projection_matrix(p):

    k, r, t, _, _, _, _ = cv2.decomposeProjectionMatrix(p)
    t = (t / t[3])[:3]
    return k, r, t
    
```

```python
def calc_depth_map(disp_left, k_left, t_left, t_right, rectified=True):
    
    if rectified:
        b = t_right[0] - t_left[0]
    else:
        b = t_left[0] - t_right[0]
        
    f = k_left[0][0]

    ## this will add noise and places that couldn't measure disparity
    disp_left[disp_left == 0.0] = 0.1
    disp_left[disp_left == -1.0] = 0.1
    
    depth_map = np.ones(disp_left.shape)
    depth_map = f * b / disp_left
    
    return depth_map
```

Let's trace the entire journey step-by-step so the root cause of this problem makes total sense. It all starts with the physical limitations of stereo cameras and ends with how computers handle colors.

### 1. The Disparity Map's Missing Data

When your `StereoSGBM` algorithm compares the left and right images, it is looking for matching pixels. However, it will inevitably fail in certain areas:

* **The Blind Spot:** The right camera is physically offset to the right. Therefore, it literally cannot see the extreme left edge of what the left camera sees.
* **Textureless Areas:** Blank white walls or solid colors offer no features for the algorithm to match.
* **Reflections:** Glass or puddles look completely different from two separate angles.

When the algorithm fails to find a match, it doesn't just leave the pixel blank; it assigns a value of `0.0` or `-1.0` to indicate an "invalid" disparity.

### 2. The Mathematical Roadblock

To turn disparity into real-world depth, you must use the stereo vision formula:

$$Z = \frac{f \cdot B}{d}$$

Because disparity ($d$) is in the denominator, you face a major programming problem. If your code tries to calculate the depth for one of those invalid pixels where $d = 0$, Python will throw a `ZeroDivisionError` and your entire program will crash.

### 3. The "Division Hack"

To keep the program running, you added a clever safety net in your code:
`disp_left[disp_left == 0.0] = 0.1`

This tells the computer: *"If you find a 0, pretend it is a 0.1 instead."* Because $0.1$ is a valid, non-zero number, the math completes successfully and your program doesn't crash.

### 4. The Consequence: Math Explosions and Visual Ruin

While the hack stops the crash, it completely destroys your data's scale.

Because depth and disparity are inversely proportional, dividing by a tiny fraction creates a massive number. For example, if $f \cdot B$ equals 380, dividing it by a normal disparity of `50` pixels gives a real depth of `7.6` meters. But dividing 380 by your hacked `0.1` results in a depth of **3800 meters**.

This triggers a chain reaction that ruins your visualization:

1. Your array is now littered with values of 3800 (representing the blind spot and scattered noise).
2. When you call `plt.imshow()`, Matplotlib automatically stretches its color scale to fit the highest number in the array.
3. It assigns bright yellow to 3800, and scales everything down to 0.
4. Because your *real* scene depths only range from roughly 1 to 50 meters, they are all squashed into the very bottom 1% of the color scale.

As a result, your real scene becomes a dark, indistinguishable purple blob, completely overshadowed by the math errors you accidentally created.

```python
import numpy as np

# This hides ANY pixel where the depth calculation blew up past 50, 
# whether it is on the left edge or a random speck in the middle.
masked_depth = np.ma.masked_where(depth > 50.0, depth) # a numpy masked array

plt.figure(figsize=(11,7))
# Because the bad data is masked in the array itself, 
# matplotlib will automatically scale the colors perfectly!
plt.imshow(masked_depth, cmap='plasma') 
plt.colorbar(label='Depth (meters)')
plt.show()
```

```python
# Additional

##############################################
### A better depth calculation function ######
##############################################
def calc_depth_map_new(disp_left, k_left, t_left, t_right, rectified=True):
    
    if rectified:
        b = t_right[0] - t_left[0]
    else:
        b = t_left[0] - t_right[0]
        
    f = k_left[0][0]
    
    # Create a mask of the valid data (where disparity is greater than 0)
    valid_pixels = disp_left > 0
    
    # Create an empty depth map filled with zeros
    depth = np.zeros_like(disp_left)
    
    # Only calculate depth WHERE the pixels are valid
    depth[valid_pixels] = (f * b) / disp_left[valid_pixels]

    masked_depth = np.ma.masked_where(depth == 0, depth)
    
    return masked_depth # returns a numpy masked array



disp_new = compute_left_disparity_map(handler.first_image_left,
                                  handler.first_image_right,
                                  matcher='sgbm',
                                  verbose=True)

depth_new = calc_depth_map_new(disp_new, k_left, t_left, t_right)

plt.figure(figsize=(11,7))
plt.grid(False)
plt.imshow(depth_new, cmap='plasma')
plt.colorbar(label='Depth (meters)')
plt.show()


plt.hist(depth_new.flatten())
plt.show()
```

```python
# Combined function

def stereo_2_depth(img_left, img_right, P0, P1, matcher='bm', rgb=False, verbose=False,
                   rectified=True):
    # Compute disparity map
    disp = compute_left_disparity_map(img_left,
                                      img_right,
                                      matcher=matcher,
                                      rgb=rgb,
                                      verbose=verbose)
    # Decompose projection matrices
    k_left, r_left, t_left = decompose_projection_matrix(P0)
    k_right, r_right, t_right = decompose_projection_matrix(P1)
    
    # Calculate depth map for left camera
    depth = calc_depth_map(disp, k_left, t_left, t_right)
    
    return depth
```

```python
# Make pointcloud homogeneous (X, Y, Z, 1)
pcloud_homo = np.hstack([pcloud[:, :3], np.ones(pcloud.shape[0]).reshape((-1,1))])
pcloud_homo
```

This single line of code is converting your standard 3D spatial coordinates into **homogeneous coordinates**.

It takes your `[X, Y, Z, Reflectance]` data, throws away the reflectance, and replaces it with a constant `1` to create rows that look like `[X, Y, Z, 1]`.

Here is exactly how the code works and why this mathematical trick is necessary for point clouds.

### The Code Breakdown

1. **`pcloud[:, :3]`**: This slices the original array, keeping all rows (`:`) but only the first three columns (`:3`). This isolates just the `X, Y, Z` coordinates, effectively dropping the reflectance value we talked about earlier.
2. **`np.ones(pcloud.shape[0])`**: `pcloud.shape[0]` returns the total number of points (e.g., `124668`). `np.ones()` then creates a 1D array of that exact length, filled entirely with `1`s.
3. **`.reshape((-1,1))`**: This takes that flat 1D array of `1`s and stands it up vertically, turning it into a single column with a shape of `(124668, 1)`.
4. **`np.hstack([...])`**: "Horizontal stack" glues these two arrays together side-by-side. It attaches the column of `1`s to the right side of the `X, Y, Z` coordinates.

### The "Why": The Translation Problem

In autonomous driving datasets like KITTI, you constantly need to transform points from one perspective to another﷿﷿﷿for example, projecting the 3D LiDAR points onto a 2D camera image.

To move a point cloud from the LiDAR's coordinate system to the camera's coordinate system, you need to do two things:

1. **Rotate it:** (Align the axes)
2. **Translate it:** (Slide the points over, because the camera is physically mounted a few inches away from the LiDAR).

In linear algebra, rotating 3D points is easy: you multiply them by a 3x3 matrix. But **you cannot translate points using 3x3 matrix multiplication**. Translation requires *addition*.

If you had to do multiplication for rotation and then addition for translation for 124,000 points, it would be computationally slow.

### The Homogeneous Solution

To combine both rotation and translation into one single, lightning-fast matrix multiplication step, computer vision relies on a 4x4 matrix.

Because you cannot multiply a 3-element `[X, Y, Z]` vector by a 4x4 matrix, we add a dummy `1` to the end of every point. This allows the math to work out perfectly:

$$\begin{bmatrix} X_{camera} \\ Y_{camera} \\ Z_{camera} \\ 1 \end{bmatrix} =  \begin{bmatrix}  R_{11} & R_{12} & R_{13} & T_x \\  R_{21} & R_{22} & R_{23} & T_y \\  R_{31} & R_{32} & R_{33} & T_z \\  0 & 0 & 0 & 1  \end{bmatrix}  \begin{bmatrix} X_{lidar} \\ Y_{lidar} \\ Z_{lidar} \\ 1 \end{bmatrix}$$

* The $R$ values handle the rotation.
* The $T$ values handle the translation.
* The appended `1` acts as a mathematical hook that grabs those $T$ values and adds them to the final coordinates during the matrix multiplication.

By running that single line of NumPy code, your point cloud is now mathematically ready to be multiplied by the KITTI sensor calibration matrices.

```python
def pointcloud2image(pointcloud, imheight, imwidth, Tr, P0):

    pointcloud = pointcloud [pointcloud[:, 0]>0]
    reflectance = pointcloud[:, 3]
    pointcloud = np.hstack([pointcloud[:, :3], np.ones(pointcloud.shape[0]).reshape((-1,1))])

    cam_xyz = Tr.dot(pointcloud.T)
    cam_xyz = cam_xyz[:, cam_xyz[2] > 0] ## Keep all 3 rows (X, Y, and Z). Only keep the columns (points) where cam_xyz[2] > 0 (z is positive)
    depth = cam_xyz[2].copy()
    cam_xyz /= cam_xyz[2]
    cam_xyz = np.vstack([cam_xyz, np.ones(cam_xyz.shape[1])])

    projection = P0.dot(cam_xyz)

    pixel_cordinates = projection.T.round(0).astype('int')

    indices = np.where((pixel_cordinates[:,0] < imwidth)
                      & (pixel_cordinates[:, 0] >= 0)
                      & (pixel_cordinates[:, 1] < imheight)
                      & (pixel_cordinates[:, 1] >=0))

    pixel_coordinates = pixel_cordinates[indices]
    depth = depth[indices]
    reflectance = reflectance[indices]

    render = np.zeros((imheight, imwidth))
    for j, (u, v, _) in enumerate(pixel_coordinates):
        if u >= imwidth or u < 0:
            continue
        if v >= imheight or v < 0:
            continue
        render[v, u] = depth[j]
        
    return render
```

### Feature extraction and matching 

```python
def extract_features(image, detector='sift', mask=None):
    
    if detector == 'sift':
        det = cv2.SIFT_create()
    elif detector == 'orb':
        det = cv2.ORB_create()
        
    kp, des = det.detectAndCompute(image, mask)
    
    return kp, des
```

```python
def match_features(des1, des2, matching='BF', detector='sift', sort=False, k=2):
    
    if matching == 'BF': # Brute-Force (BF) Matcher
        if detector == 'sift':
            matcher = cv2.BFMatcher_create(cv2.NORM_L2, crossCheck=False)
        elif detector == 'orb':
            matcher = cv2.BFMatcher_create(cv2.NORM_HAMMING2, crossCheck=False)
    elif matching == 'FLANN': # Fast Library for Approximate Nearest Neighbors
        FLANN_INDEX_KDTREE = 1
        index_params = dict(algorithm=FLANN_INDEX_KDTREE, trees=5)
        search_params = dict(checks=50)
        matcher = cv2.FlannBasedMatcher(index_params, search_params)
    
    matches = matcher.knnMatch(des1, des2, k=k) # k = 2 for a feature in the first image we need 2 matches for t+1
        
    if sort:
        matches = sorted(matches, key=lambda x: x[0].distance)
        
    return matches
```

```python
def visualize_matches(image1, kp1, image2, kp2, match):
    image_matches = cv2.drawMatches(image1, kp1, image2, kp2, match, None, flags=2)
    plt.figure(figsize=(16, 6), dpi=100)
    plt.imshow(image_matches)
```

```python
def filter_matches_distance(matches, dist_threshold=0.5):
    filtered_matches = []
    for m, n in matches:
        if m.distance <= dist_threshold * n.distance:
            filtered_matches.append(m)
            
    return filtered_matches
```

```python
def estimate_motion(matches, kp1, kp2, k, depth1, marx_depth=3000):
    rmat = np.eye(3)
    tvec = np.zeros((3, 1))

    image1_points = np.float32([kp1[m.queryIdx].pt for m in matches])
    image2_points = np.float32([kp2[m.trainIdx].pt for m in matches])

    cx = k[0, 2]
    cy = k[1, 2]
    fx = k[0, 0]
    fy = k[1, 1]

    object_points = np.zeros((0, 3))
    delete = []

    for i, (u, v) in enumerate(image1_points):

        z = depth1[int(round(v)), int(round(u))]

        if z > max_depth:
            delete.append(i)
            continue


        x = z * (u - cx) / fx
        y = z * (v - cy) / fy

        object_points = np.vstack([object_points, np.array([x, y, z])])

        #object_points = np.vstack([obeject_points, np.linalg.inv(k).dot(z*np.array([u, v, 1]))])

    image1_points = np.delete(image1_points, delete, 0)
    image2_points = np.delete(image2_points, delete, 0)

    _, rvec, tvec, inliners = cv2.solvePnPRansac(object_points, image2_points, k , None) # Random sample concensus
    rmat = cv2.Rodrigues(rvec)[0]

    return rmat, tvec, image1_points, image2_points
    
```

```python
def visual_odometry(handler, detector='sift', matching='BF', filter_match_distance = None, 
                    stereo_matcher='sgbm', mask= None, subset=None, plot=None):

    lidar_correction = handler.lidar
    print(f'Generating diparities with stereo {stereo_matcher}')
    print(f'Detecting features with {detector} and matching with {matching}')

    if filter_match_distance is not None:
        print('Filtering feature matches at htreshold of {filter_match_distance}*distance')
    if lidar_correction:
        print('Improving stereo depth estimation with lidar data')
    if subset is not None:
        num_frames = subset
    else:
        num_frames = handler.num_frames

    if plot:
        fig = plt.figure(figsize=(14, 14))
        ax = fig.add_subplot(projection='3d')
        ax.view_init(elev=-20, azim=270)
        xs = handler.gt[:, 0, 3] # xs, ys, zs are Translation Vector, representing the physical location of the camera in 3D space.
        ys = handler.gt[:, 1, 3]
        zs = handler.gt[:, 2, 3]
        ax.set_box_aspect((np.ptp(xs), np.ptp(ys), np.ptp(zs)))
        ax.plot(xs, ys, zs, c='k')

    # Establish a homogeneous transformation matrix. First pose is identity
    T_tot = np.eye(4)  
    trajectory = np.zeros((num_frames, 3, 4))
    trajectory[0] = T_tot[:3, :]
    imheight = handler.imheight
    imwidth = handler.imwidth

    k_left, r_left, t_left = decompose_projection_matrix(handler.P0)

    if handler.low_memory:

        handler.reset_frame()
        image_plus1 = next(handler.images_left)

    for i in range(num_frames - 1):
        start = datetime.datetime.now()

        if handler.low_memory:
            image_left = image_plus1
            image_plus1 = next(handler.images_left)
            image_right = next(handler.images_right)
        else:
            image_left = handler.images_left[i]
            image_plus1 = handler.images_left[i+1]
            image_right = handler.images_right[i]

        depth = stereo_2_depth(image_left,
                               image_right,
                               P0 = handler.P0,
                               P1 = handler.P1,
                               matcher=stereo_matcher,
                               rgb=False,
                               verbose=True)

        if lidar_correction:
            if handler.low_memory:
                pointcloud = next(handler.pointclouds)
            else:
                pointcloud = handler.pointclouds[i]

            lidar_depth = pointcloud2image(pointcloud, 
                          imheight = imheight,
                          imwidth = imwidth,
                          Tr = handler.Tr,
                          P0 = handler.P0)

        indices = np.where(lidar_depth > 0)
        depth[indices] = lidar_depth[indices]


        # get keypoints and descriptors for left camera image of 2 sequetial images
        kp0, des0 = extract_features(image_left, detector, mask)
        kp1, des1 = extract_features(image_plus1, detector, mask)

        # Filter matches if a distance threshold is provided by user
        matches_unfiltered = match_features(des0,
                                            des1,
                                            matching=matching,
                                            detector=detector,
                                            sort=False)
        if filter_match_distance:
            matches = filter_matches_distance(matches_unfiltered, filter_match_distance)
        else:
            matches = matches_unfiltered

        # Estimate motion between sequential images of the left camera
        rmat, tvec, img1_points, img2_points = estimate_motion(matches,
                                                               kp0,
                                                               kp1,
                                                               k_left,
                                                               depth)

        tmat = np.hstack([rmat, tvec])
        homo_tmat = np.vstack([tmat, [0, 0 , 0 , 1]])

        T_tot = T_tot.dot(np.linalg.inv(homo_tmat))

        trajectory[i+1, :, :] = T_tot[:3, :]

        end = datetime.datetime.now()
        print('Time to compute frame {}:'.format(i+1), end-start)

        if plot:
            xs = trajectory[:i+2, 0, 3]
            ys = trajectory[:i+2, 1, 3]
            zs = trajectory[:i+2, 2, 3]
            plt.plot(xs, ys, zs, c='chartreuse')
            plt.pause(1e-32)

    if plot:
        plt.close()
        
    return trajectory
        
            
    
```
